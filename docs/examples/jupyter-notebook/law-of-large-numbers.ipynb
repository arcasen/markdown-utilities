**大数定律**

大数定律（Law of Large Numbers，简称 LLN）是概率论中最基础也最重要的定理之一。简单来说：**在重复进行的独立随机试验中，随着试验次数 $n$ 的无限增加，样本的平均值（频率）会越来越接近总体的理论期望值（概率）**。

Python 模拟：我们可以通过模拟抛硬币（正面为 1，反面为 0，理论期望为 0.5）来观察随着抛掷次数增加，正面出现的累积比例是如何平稳收敛到 0.5 的。

```python
import numpy as np
import matplotlib.pyplot as plt

# 设置随机种子
np.random.seed(42)

# 模拟抛硬币 10,000 次（1 表示正面，0 表示反面）
num_tosses = 10000
tosses = np.random.choice([0, 1], size=num_tosses, p=[0.5, 0.5])

# 计算累积平均值（即每一次抛掷后正面的累积比例）
cumulative_means = np.cumsum(tosses) / np.arange(1, num_tosses + 1)

# 绘制收敛过程
plt.figure(figsize=(10, 5))
plt.plot(cumulative_means, label='样本均值（正面比例）', color='tab:blue', linewidth=1.5)
plt.axhline(y=0.5, color='red', linestyle='--', label='理论期望 (0.5)')

plt.xscale('log')  # 使用对数坐标轴便于观察前期波动与后期平稳
plt.title('大数定律模拟：抛硬币累计正面比例')
plt.xlabel('抛掷次数 (n, 对数刻度)')
plt.ylabel('正面比例')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
